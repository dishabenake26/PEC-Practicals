```python

import pandas as pd

data = {
    "Name": ["Amit", "Disha", "Sai", "Priya", "Sneha"],
    "Age": [20, 21, 23, 18, 24],
    "City": ["Pune", "Mumbai", "Dubai", "Kolhapur", "Bangladesh"],
    "Marks": [90, 99, 98, 97, 79]
}

df = pd.DataFrame(data)

# 1. Display data
print(df)

# 2. Select a single column
print(df["Name"])

# 3. Select multiple columns
print(df[["Name", "Marks"]])

# 4. Select first 3 rows
print(df.head(3))

# 5. Filter marks greater than 80
print(df[df["Marks"] > 80])

# 6. Filter students from Pune
print(df[df["City"] == "Pune"])

# 7. Filter students aged 20
print(df[df["Age"] == 20])

# 8. Add a new column
df["Result"] = "Pass"
print(df)

# 9. Add a Grade column
df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x >= 80 else "B"
)
print(df)

# 10. Update marks
df.loc[df["Name"] == "Disha", "Marks"] = 100
print(df)

# 11. Sort by marks
print(df.sort_values("Marks", ascending=False))

# 12. Calculate average marks
print(df["Marks"].mean())

# 13. Find highest marks
print(df["Marks"].max())

# 14. Find lowest marks
print(df["Marks"].min())

# 15. Delete a column
df = df.drop("Age", axis=1)
print(df)
```
