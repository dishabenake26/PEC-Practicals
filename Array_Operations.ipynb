```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50])
print("Array:",arr)
```

```python
arr=np.array([10,20,30,40,50])
print("Array:",arr)
print("The size of the array is:",arr.size)
print("Dimension:",arr.ndim)
```

```python
arr=np.array([[10,20,30],[40,50,60]])
print("2-D Array:")
print(arr)
```

```python
a=np.array([10,20,30,40,50])
print("Array:",a)
print("Addition:",a+5)
print("Substraction:",a-5)
print("Multiplication:",a*5)
print("Division:",a/5)
print("First Element:",a[1])
print("Second Element:",a[2])
print("Last Element:",a[-1])
print("First Three Elements:",a[0:3])
print("Elements from second element:",a[2:])


```

```python
b=np.array([10,20,30,40,50])
print("Marks:",b)
print("Sum:",np.sum(b))
print("Mean:",np.mean(b))
print("Maximum:",np.max(b))
print("Minimum:",np.min(b))
```

```python
marks=np.array([60,50,40,30,20,10])
print("Marks:",marks)
print("Number of elements:",marks.size)
print("Dimension:",marks.ndim)
print("Marks+5:",marks+5)
print("Marks*2:",marks*2)
print("Total Marks:",np.sum(marks))
print("Average of the marks:",np.mean(marks))
print("The maximum marks:",np.max(marks))
print("The minimum mark:",np.min(marks))
```
