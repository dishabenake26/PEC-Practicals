```python

import numpy as np

# Create an array
arr = np.array([1, 4, 9, 16, 25])

print("Original Array:", arr)

# 1. Square Root
print("Square Root:", np.sqrt(arr))

# 2. Square
print("Square:", np.square(arr))

# 3. Addition
print("Addition:", np.add(arr, 5))

# 4. Subtraction
print("Subtraction:", np.subtract(arr, 2))

# 5. Multiplication
print("Multiplication:", np.multiply(arr, 3))

# 6. Division
print("Division:", np.divide(arr, 2))

# 7. Exponential
print("Exponential:", np.exp(arr))

# 8. Logarithm
print("Logarithm:", np.log(arr))

# 9. Maximum
print("Maximum:", np.maximum(arr, 10))

# 10. Minimum
print("Minimum:", np.minimum(arr, 10))
```
